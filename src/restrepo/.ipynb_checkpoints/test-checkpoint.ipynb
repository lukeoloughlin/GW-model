```python
import math
import os

#os.environ['NUMBA_ENABLE_CUDASIM']='1'

import numpy as np
import numba
import numba.cuda as cuda
from numba import void, f4
from numba.cuda.random import create_xoroshiro128p_states

from CRU import CRUs, RyR_stationary, LCC_stationary, DEFAULT_RYR_PARMS
from RyR import test_RyR_cpu, RyR_kernel, update_RyR_rates
```

```python
cp = 10 * np.ones((200, 200), dtype=np.float32)
```

```python
RyR = RyR_stationary(cp)
```

```python
@cuda.jit(
    #void(
    #    f4[:, :, :],
    #    f4[:, :, :],
    #    f4[:, :],
    #    f4,
    #    f4,
    #    f4,
    #    f4,
    #    f4,
    #    f4,
    #    f4,
    #    f4,
    #),
)
def update_RyR_rates_(
    RyR_rates,
    RyR,
    cp,
    Ku,
    Kb,
    _1_tau_u,
    _1_tau_b,
    _1_tau_c,
    BCSQN,
    rho_inf,
    K,
):
    x, y = cuda.grid(2)
    N1, N2, _ = RyR.shape
    if x < N1 and y < N2:
        update_RyR_rates(RyR_rates, RyR, cp, Ku, Kb, _1_tau_u, _1_tau_b, _1_tau_c, BCSQN, rho_inf, K, x, y)
```

```python
RyR_ = np.copy(RyR)
tpb_RyR = 16


d_RyR = cuda.to_device(RyR_)
d_cp = cuda.to_device(cp)
RyR_rates = cuda.device_array((RyR_.shape[0], RyR_.shape[1], 8), dtype=np.float32)

Ku = np.float32(DEFAULT_RYR_PARMS["Ku"])
Kb = np.float32(DEFAULT_RYR_PARMS["Kb"])
tau_u = np.float32(DEFAULT_RYR_PARMS["tau_u"])
tau_b = np.float32(DEFAULT_RYR_PARMS["tau_b"])
tau_c = np.float32(DEFAULT_RYR_PARMS["tau_c"])
BCSQN = np.float32(DEFAULT_RYR_PARMS["BCSQN"])
rho_inf = np.float32(DEFAULT_RYR_PARMS["rho_inf"])
K = np.float32(DEFAULT_RYR_PARMS["K"])
_1_tau_u = 1 / tau_u
_1_tau_b = 1 / tau_b
_1_tau_c = 1 / tau_c
```

```python
bpg_RyR = (math.ceil(RyR_.shape[0] / tpb_RyR),
        math.ceil(RyR_.shape[1] / tpb_RyR),
        )
```

```python
update_RyR_rates_[bpg_RyR, (16, 16)](RyR_rates, d_RyR, d_cp, Ku, Kb, np.float32(_1_tau_u), np.float32(_1_tau_b), np.float32(_1_tau_c), BCSQN, rho_inf, K)
cuda.synchronize()
rates_cpu = RyR_rates.copy_to_host()
```

```python
Ku = np.float32(DEFAULT_RYR_PARMS["Ku"])
Kb = np.float32(DEFAULT_RYR_PARMS["Kb"])
tau_u = np.float32(DEFAULT_RYR_PARMS["tau_u"])
tau_b = np.float32(DEFAULT_RYR_PARMS["tau_b"])
tau_c = np.float32(DEFAULT_RYR_PARMS["tau_c"])
BCSQN = np.float32(DEFAULT_RYR_PARMS["BCSQN"])
rho_inf = np.float32(DEFAULT_RYR_PARMS["rho_inf"])
K = np.float32(DEFAULT_RYR_PARMS["K"])
_1_tau_u = 1 / tau_u
_1_tau_b = 1 / tau_b
_1_tau_c = 1 / tau_c

RyR_ = np.copy(RyR)
tpb_RyR = 16
dt = 5e-3

sqrtdt = np.sqrt(dt, dtype=np.float32)

d_RyR = cuda.to_device(RyR_)
d_cp = cuda.to_device(cp)
dW = cuda.device_array((RyR_.shape[0], RyR_.shape[1], 4), dtype=np.float32)
RyR_rates = cuda.device_array((RyR_.shape[0], RyR_.shape[1], 8), dtype=np.float32)
RyR_sorted = cuda.device_array_like(RyR_)
        
rng_states = create_xoroshiro128p_states(RyR_.shape[0] * RyR_.shape[1], seed=0)
cuda.synchronize()


bpg_RyR = (math.ceil(RyR_.shape[0] / tpb_RyR),
        math.ceil(RyR_.shape[1] / tpb_RyR),
        )
RyR_kernel[bpg_RyR, (tpb_RyR, tpb_RyR)](
d_RyR,
RyR_rates,
RyR_sorted,
d_cp,
dW,
np.float32(0.1),
np.float32(dt),
sqrtdt,
Ku,
Kb,
_1_tau_u,
_1_tau_b,
_1_tau_c,
BCSQN,
rho_inf,
K,
rng_states,
)
cuda.synchronize()
d_RyR.copy_to_host(RyR_)
```
